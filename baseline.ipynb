```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
train_data = pd.read_csv("train.csv")
test_data = pd.read_csv("test.csv")
```

```python
y_train = train_data["Class Index"]
y_test = test_data["Class Index"]

X_train = train_data.drop(["Class Index"], axis = 1)
X_train["text"] = X_train["Title"].fillna("")+" "+X_train["Description"].fillna("")

X_test = test_data.drop(["Class Index"], axis = 1)
X_test["text"] = X_test["Title"].fillna("")+" "+X_test["Description"].fillna("")
```

```python
vectorizer = TfidfVectorizer(max_features=10000)
TfidX_train = vectorizer.fit_transform(X_train["text"])
TfidX_test = vectorizer.transform(X_test['text'])
```

```python
clf = LinearSVC(random_state=0, max_iter=2000).fit(TfidX_train, y_train)
```

```python
y_pred_test = clf.predict(TfidX_test)
```

```python
accuracy_score(y_test, y_pred_test)
```

```python
print(classification_report(y_test, y_pred_test, target_names = ['Мир', 'Спорт', 'Бизнес', 'Наука/Технологии']))
```

```python
feature_name = vectorizer.get_feature_names_out()
for i, class_label in enumerate(clf.classes_):
    top10 = feature_name[clf.coef_[i].argsort()[-10:]]
    print(f"Класс {class_label}: {list(top10)}")
```
